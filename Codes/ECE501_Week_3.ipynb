```python

import pandas as pd

germany = pd.read_csv(
    "/content/germany_final_mixed_50_rows_v4.csv"
)

uruguay = pd.read_csv(
    "/content/uruguay_final_mixed_50_rows_v4.csv"
)

print("GERMANY")
print("Shape:", germany.shape)
print("Columns:", germany.columns.tolist())
display(germany.head())

print("\nURUGUAY")
print("Shape:", uruguay.shape)
print("Columns:", uruguay.columns.tolist())
display(uruguay.head())
```

```python

# Check important columns and missing values

print("GERMANY COLUMNS")
print(germany.columns.tolist())

print("\nURUGUAY COLUMNS")
print(uruguay.columns.tolist())

print("\nGERMANY DATA TYPES")
print(germany.dtypes)

print("\nURUGUAY DATA TYPES")
print(uruguay.dtypes)

print("\nGERMANY MISSING VALUES")
print(germany.isnull().sum())

print("\nURUGUAY MISSING VALUES")
print(uruguay.isnull().sum())
```

```python

# Display possible crop labels, IDs, coordinates and dates

keywords = [
    "crop", "label", "target", "field", "index",
    "latitude", "longitude", "lat", "lon",
    "coordinate", "date", "time", "country"
]

for name, df in [
    ("GERMANY", germany),
    ("URUGUAY", uruguay)
]:
    print("\n==========", name, "==========")

    for col in df.columns:
        if any(word in col.lower() for word in keywords):
            print(f"\nColumn: {col}")
            print("Unique values:", df[col].nunique(dropna=True))
            print(df[col].dropna().unique()[:10])
```

```python

for name, df in [("Germany", germany), ("Uruguay", uruguay)]:
    print(f"\n{'='*12} {name} {'='*12}")

    print("\nCrop, target and field mapping:")

    display(
        df.groupby(
            ["crop", "target", "farm_identifier",
             "field_shared_name"],
            dropna=False
        ).size().reset_index(name="rows")
    )

    print("\nBand-wise values:")

    display(
        df.groupby("band")["stats-mean"]
        .agg(["count", "mean", "min", "max"])
        .reset_index()
        .head(20)
    )

    print("\nSample rows:")
    display(
        df[["crop", "target", "crop_t", "band",
            "time_step", "year", "times",
            "stats-mean"]].head(12)
    )
```

```python

for name, df in [("Germany", germany), ("Uruguay", uruguay)]:
    print(f"\n===== {name} =====")

    print("\n1. Crop and target mapping:")
    display(
        df[["crop", "target", "crop_t"]]
        .drop_duplicates()
        .sort_values("crop")
    )

    print("\n2. Bands available:")
    print(sorted(df["band"].dropna().astype(str).unique()))

    print("\n3. Rows per crop and band:")
    display(
        df.groupby(["crop", "band"])
        .size()
        .unstack(fill_value=0)
    )

    print("\n4. Missing values in important columns:")
    cols = [
        "crop", "target", "crop_t", "band",
        "time_step", "year", "times", "stats-mean"
    ]
    print(df[cols].isna().sum())
```

```python

optical_bands = [
    "B02", "B03", "B04", "B05",
    "B06", "B07", "B08", "B8A",
    "B11", "B12"
]

for name, df in [("Germany", germany), ("Uruguay", uruguay)]:
    available = sorted(set(df["band"].dropna()) & set(optical_bands))

    print(f"\n{name}")
    print("Available optical bands:", available)
    print("Number of optical bands:", len(available))
    print("Unique fields:",
          df[["farm_identifier", "field_shared_name"]]
          .drop_duplicates().shape[0])
    print("Unique time steps:", df["time_step"].nunique())
    print("Unique years:", df["year"].dropna().unique())
```

```python

for name, df in [("Germany", germany), ("Uruguay", uruguay)]:
    print(f"\n===== {name} =====")

    bands = df["band"].dropna().astype(str).unique()

    print("SAR VV available:", "VV" in bands)
    print("SAR VH available:", "VH" in bands)

    print("\nOptical band statistics:")
    display(
        df[df["band"].isin([
            "B02", "B03", "B04", "B05", "B06",
            "B07", "B08", "B8A", "B11", "B12"
        ])][["crop", "year", "band", "stats-mean"]]
        .sort_values(["crop", "year", "band"])
        .head(30)
    )

    print("\nRows per year:")
    print(df.groupby("year").size())
```

```python

import pandas as pd
import matplotlib.pyplot as plt

# Optical bands available in both datasets
optical_bands = [
    "B02", "B03", "B04", "B05", "B06",
    "B07", "B08", "B8A", "B11", "B12"
]

# Graph 1: Crop distribution
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

germany["crop"].value_counts().plot(
    kind="bar", ax=axes[0]
)
axes[0].set_title("Germany: Crop Distribution")
axes[0].set_xlabel("Crop")
axes[0].set_ylabel("Number of Rows")
axes[0].tick_params(axis="x", rotation=0)

uruguay["crop"].value_counts().plot(
    kind="bar", ax=axes[1]
)
axes[1].set_title("Uruguay: Crop Distribution")
axes[1].set_xlabel("Crop")
axes[1].set_ylabel("Number of Rows")
axes[1].tick_params(axis="x", rotation=0)

plt.tight_layout()
plt.show()


# Graph 2: Mean values of optical bands
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

for df, country, ax in [
    (germany, "Germany", axes[0]),
    (uruguay, "Uruguay", axes[1])
]:
    optical = df[df["band"].isin(optical_bands)]

    band_means = optical.groupby("band")["stats-mean"].mean()

    band_means.plot(kind="bar", ax=ax)
    ax.set_title(country + ": Optical Band Mean Values")
    ax.set_xlabel("Optical Band")
    ax.set_ylabel("Mean Pixel Value")
    ax.tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt

optical_bands = [
    "B02", "B03", "B04", "B05", "B06",
    "B07", "B08", "B8A", "B11", "B12"
]

fig, axes = plt.subplots(1, 2, figsize=(16, 5))

for df, country, ax in [
    (germany, "Germany", axes[0]),
    (uruguay, "Uruguay", axes[1])
]:
    data = df[df["band"].isin(optical_bands)]

    comparison = data.pivot_table(
        index="band",
        columns="crop",
        values="stats-mean",
        aggfunc="mean"
    )

    comparison.plot(kind="bar", ax=ax)

    ax.set_title(country + ": Optical Bands by Crop")
    ax.set_xlabel("Optical Band")
    ax.set_ylabel("Average stats-mean")
    ax.tick_params(axis="x", rotation=45)
    ax.legend(title="Crop")

plt.tight_layout()
plt.show()
```
